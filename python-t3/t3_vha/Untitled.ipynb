```python
url="http://127.0.0.1:8000/api/players/"
res=requests.get(url)
print(res.json())
print(res.status_code)
```

```python
header={
    "Authorization": "Token 176f17530fdbb5308f3dbc1419100c53f4ecdaa8"
}
url="http://127.0.0.1:8000/api/players/"
res=requests.get(url,headers=header)
print(res.json())
print(res.status_code)
```

```python
data={"name": "A",
      "runs": 1200,
      "innings": 275}
url="http://127.0.0.1:8000/api/players/"
res=requests.post(url,json=data)
print(res.json())
```

```python
print(res.status_code)
```

```python
data={"name": "B",
      "runs": 1200,
      "innings": 275}
url="http://127.0.0.1:8000/api/players/"
res=requests.put(url,json=data)
print(res.json())
```

```python
data={
      "innings": 400}
url="http://127.0.0.1:8000/api/players/9/"
res=requests.patch(url,json=data)
print(res.json())
```

```python

url="http://127.0.0.1:8000/api/players/9/"
res=requests.delete(url)

print(res.status_code)
```
